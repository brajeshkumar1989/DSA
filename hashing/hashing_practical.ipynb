# **Hashing Dictionary**

Dictionary creation using Hasing 

* Maintain 2 arrays one for keys and another is for data (both wil always have same size)







Chaining means create node chain for a perticular index position of array

probing means go to next available index position

1. Linear Probing

Create an array of given size

Define a hash function

```python
class Dictionary:
    def __init__(self, size):
        self.size = size  # size of array
        # two arrays with value None
        self.slots = [None] * self.size
        self.data = [None] * self.size

    def put(self, key, value):
        # first calculate hash value
        hash_val = self.hasn_functin(key)
        # when index position has None or not None
        if self.slots[hash_val] == None:
            self.slots[hash_val] = key  # key array
            self.data[hash_val] = value  # data array

        else:
            # when index value is not None
            # when same key , or different key
            if self.slots[hash_val] == key:
                self.data[hash_val] = value
            else:
                # now same key is not there but for different key has same hash_value, and now i cannot change previous key and then i will move to next index (using linear probing)
                # and exact on same position i will have to update the data in 2nd value array
                # when i use Linear probing it means i will go to next index (move right hand side) and now i need to calculate new hash value and for that i will use rehash function to increase with val 1
                new_hash_value = self.rehash(hash_val)
                # there is again two possibilites , new index position is empty or not empty
                # second condition in while loop is when the key already available.
                while (
                    self.slots[new_hash_value] != None
                    and self.slots[new_hash_value] != key
                ):  # slot is representing a key -> array
                    new_hash_value = self.rehash(new_hash_value)

                if self.slots[new_hash_value] == None:
                    self.slots[new_hash_value] = key
                    self.data[new_hash_value] = value
                else:
                    self.data[new_hash_value] = (
                        value  # when self.slots[new_hash_value] == key) then update the current value with new value.
                    )
                    
    def get(self, key):
        
        # calculate the hashvalue
        start_position = hash(key)

    def __setitem__(
        self, key, value
    ):  # so now i can use now object with D1["name"] = value kind of call
        self.put(key, value)

    def rehash(self, old_hash):
        return (old_hash + 1) % self.size

    def hasn_functin(self, key):
        # the main reason of using python hash is , we can find modulus for integers , but in real case key can be string as well, so calculating hash gives a long number (and it is always going to be a constant number for the same key)
        return abs(hash(key)) % self.size
```
