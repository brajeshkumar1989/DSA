# **Hashing**

Why Hashing?

It is a fast way to search.

Types

1. Linear Search -> I start from the beginning of the array and then i break the loop once I find the required value. O(n) , if we double the array it also double the time required to search.

2. Binary Search -> first need to sort the array and then i take the middle one and then i compare with the element i want to search and accordingly split the array and based on left or right i use that sub array and again repeat the same (finding middle one and go left and right based on comparision with the element i want to search). And time complexity is O(log n)

3. Better than O(log n) is O(1) and this is what hashing gives us.

***
In case of array to find the value of a array[i] , time complexity is O(1) by using formula 
initial_address + Index_Position * size of one field (1 byte or some bits)

and here i am not using length of the list and that is the reason it is O(1)

Example:
Let I have array with size 5, and i want to store some elements {0,1,2,3,4} and i want to store it in a way so that searching time complexity can be reduced to O(1).

if i store value to its index position means 0 stores at 0 , 5 store at 5 (this is what hashing is)

Problem is there in this approach and if i have {0,1,3,10000} means for these 4 value to store I need 10000 size of array -> Wastage of storage.

Now will use new function and it is known as Hashing function.

Let's say I have 10 size of array and i want to store {11,22,33,44,55,66,77,88,99,100}
simple hash function : i % size

11 % 10  = 1 , and it will store at position 1
22 % 10 = 2 , and it will store at position 2
100 % 10 = 0 , and it will store at position 0

Now when i want to find 100 then i will pass that value 100 to hash function and it will give me 0 and then that is the index position of that element in array.

if I want to find 101 , so 101 % 10 = 1, it means i will check index positon 1 and then there I find 11 is stored so i match 11 with 101 which is False and i simply return 101 doesn't exist.

Linear Probing is like i will start from the position and go through all the index position by using (g(i) = [h(i) + k(i')] % size) hash function , will keep moving right and then come back to 0 index and go till the start position - 1

Problem with Linear Probing , it creates clustering kind of detail and keep other memory place empty, to avoidn this issue there is a solution known as **Quadratic Probing**

**g(i)=[h(i) + k(i')] % size**, where everything is same except K(i') is a quadratic means k(i')= (i')^2 where i'=0,1,2,3,4...



It helps to avoid clustering issue and utilize more array space.
