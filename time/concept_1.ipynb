## Orders of Growth

Goal:

* Want to evaluate program's efficiency when **input is very big**.
* Want to express the **growth of program's run time** as input size grows. - Mathematical formula between time and input (graph)
* Want to put an **upper bound** on growth - as tigh as possible.
* Do not need to be precise: **"order of" not "exact"** growth. - `i^2+2i+3 means i^2`
* We will look at **largest factors** in run time (which section of program will take the longest to run?) - nested for loop will take more time as compare with for loop
* **Thus, generally we want tight upper bound on growth, as function of size of input, in worst case**.


### Measuring Order of Growth: BIG Oh Notation

* Big Oh notation measures an **upper bound on the asymptitic growth**, often called order of growth.

* **Big Oh or O()** is used to describe worst case.
    - Worst case occurs often and is the bottleneck when a program runs.
    - Express rate of growth of program relative to the input size.
    - Evaluate algorithm **NOT** machine or implementation.

Idea is simple

* n^2 + 2n + 2 , where square means nested operation, 2n means a loop, and single 2 is 2 operations outside of loop , so O(n^2)

* n^2+100000n+3^1000 , o(n^2)

* log(n) + n + 4 , o(n)

* 0.0001 * n * log(n) + 300n , o(nlogn)

* 2n^30+ 3^n, o(3^n)




**Constant**: in case of Array for a specific index element i access even when increasing elements for for that specific index it will always take same time (t sec)
1st memory address + 4 X index (4 is the memory) 

**Linear**: input and time propotional to each other. (Linear search , increasing array length increases the search time)

**Quadratic**: Not good (like nested loops) (Booble sort)

**Logaritmic**: input increases 10 times and time increases 1 time. - After constant best one
example: Binary search

**nlogn**: bad as compare with **Linear** but better than **Quadratic**.
example: Merge sort

**exponential**: Worse,

```python
def fact_iterative(n):
    """
    Total operations =5n because of loop and 4 operations inside the loop. and 1 operation outside the loop. So, total operations = 5n + 1 so order of growth = O(n) means linear time complexity.
    """
    
    result = 1 
    while n > 1: # operation 1
        result *= n # operation 2 and 3 (multiplication and assignment)
        n -= 1 # operation 4 (decrement)
    return result
```

```python
# Problem 3 (Linear Search)

# Problem 4 , input increasing 123, 1234 , and in below code everytime we are dividing the input by 10 so the time complexity is O(log10(n)) since we are reducing the input size by a factor of 10 in each iteration.

def intToStr(i):
    """
    This is a function that converts an integer to a string representation. The function takes an integer as input and returns its string representation. The time complexity of this function is O(log10(i)) since the number of digits in the integer determines the number of iterations in the while loop.
    """
    
    digits = '0123456789'
    if i == 0:
        return '0'
    result = ''
    while i > 0:
        result = digits[i % 10] + result
        i //= 10
    return result
    
print(intToStr(123))
```

```python
# Problem 12

def factorial(n): # O(n) time complexity
    if n == 1 or n == 0:
        return 1
    return n * factorial(n-1)
print(factorial(5))

# input is 5 and the function calls itself 5 times until it reaches the base case of n=1 or n=0. so the time complexity is O(n) since the number of recursive calls is proportional to the input size n.
```

```python
# Problem 13
# when input is 1 then function call is 1 and n = 2 , function call is 2 , 3-> 4, 5-> 14, 6-> approx. 32 , then it is exponential time complexity O(2^n) since the number of function calls doubles with each increase in input size n. The time complexity can be derived from the recurrence relation T(n) = T(n-1) + T(n-2) + O(1), which is similar to the Fibonacci sequence. The number of function calls grows exponentially with respect to the input size n.
# what is the time complexity of the following function?
# it is less than O(2^n) and proven in research that the time complexity is approx. O(1.7^n) and can be optimized to O(n) using dynamic programming or memoization techniques.
def fib(n):
    if n==1 or n==0:
        return 1
    return fib(n-1) + fib(n-2)

```

```python
# problem 14

def power(num):
    if num < 1:
        return 0
    elif num == 1:
        print(1)
        return 1
    else:
        prev = power(num // 2)
        curr = prev * 2
        print(curr)
        return curr

power(145)

# This program complexity is O(log n) since we are dividing the input size by 2 in each iteration. when n=100 then num=100,50,25,12,6,3,1 so 7 iterations. when n=1000 then num=1000,500,250,125,62,31,15,7,3,1 so 10 iterations. so the time complexity is O(log n).
```

```python
# problem 15

def mod(a, b):
    if b <= 0:
        return -1
    div = a // b
    return a - div * b


# This function calculates the modulus of two numbers.Time complexity of this function is O(1) since it performs a constant number of operations regardless of the input size.

mod(5, 3)
```

```python
# problem 16

def sum_digits(num):
    sum = 0
    while num > 0:
        sum += num % 10
        num //= 10
    return sum

# This function calculates the sum of digits of a number. The time complexity of this function is O(log10(n)) since the number of digits in the integer determines the number of iterations in the while loop.
sum_digits(1234)
```
