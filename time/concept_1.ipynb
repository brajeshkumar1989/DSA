## Orders of Growth

Goal:

* Want to evaluate program's efficiency when **input is very big**.
* Want to express the **growth of program's run time** as input size grows. - Mathematical formula between time and input (graph)
* Want to put an **upper bound** on growth - as tigh as possible.
* Do not need to be precise: **"order of" not "exact"** growth. - `i^2+2i+3 means i^2`
* We will look at **largest factors** in run time (which section of program will take the longest to run?) - nested for loop will take more time as compare with for loop
* **Thus, generally we want tight upper bound on growth, as function of size of input, in worst case**.


### Measuring Order of Growth: BIG Oh Notation

* Big Oh notation measures an **upper bound on the asymptitic growth**, often called order of growth.

* **Big Oh or O()** is used to describe worst case.
    - Worst case occurs often and is the bottleneck when a program runs.
    - Express rate of growth of program relative to the input size.
    - Evaluate algorithm **NOT** machine or implementation.

Idea is simple

* n^2 + 2n + 2 , where square means nested operation, 2n means a loop, and single 2 is 2 operations outside of loop , so O(n^2)

* n^2+100000n+3^1000 , o(n^2)

* log(n) + n + 4 , o(n)

* 0.0001 * n * log(n) + 300n , o(nlogn)

* 2n^30+ 3^n, o(3^n)




**Constant**: in case of Array for a specific index element i access even when increasing elements for for that specific index it will always take same time (t sec)
1st memory address + 4 X index (4 is the memory) 

**Linear**: input and time propotional to each other. (Linear search , increasing array length increases the search time)

**Quadratic**: Not good (like nested loops) (Booble sort)

**Logaritmic**: input increases 10 times and time increases 1 time.

```python
def fact_iterative(n):
    """
    Total operations =5n because of loop and 4 operations inside the loop. and 1 operation outside the loop. So, total operations = 5n + 1 so order of growth = O(n) means linear time complexity.
    """
    
    result = 1 
    while n > 1: # operation 1
        result *= n # operation 2 and 3 (multiplication and assignment)
        n -= 1 # operation 4 (decrement)
    return result
```
