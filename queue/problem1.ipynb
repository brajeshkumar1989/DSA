````python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None
````

````python
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    def enqueue(self, value):

        new_node = Node(value)

        if self.rear == None:
            self.front = new_node
            self.rear = self.front
        
        else:
            self.rear.next = new_node
            self.rear = new_node

    def dequeue(self):
        if self.front == None:
            return "Empty"
        else:
            self.front = self.front.next

    def isempty(self):
        return self.front == None

    def size(self):
        temp = self.front
        counter = 0
        while temp != None:
            counter += 1
            temp = temp.next
        return counter
    
    def front_item(self): #peak
        
        if self.front == None:
            
            return "Empty"
        
        else:
            return self.front.data
        
        
    def rear_item(self): #peak from rear
        
        if self.rear == None:
            
            return "Empty"
        
        else:
            return self.rear.data    
                  

    def traverse(self):
        temp = self.front

        while temp != None:
            print(temp.data, end=" ")
            temp = temp.next
````

````python
def fun(num):
    if num == 0:
        return 0
    else:
        queue.enqueue(num % 10)
        res = fun(num // 10)
        res = res * 10 + queue.dequeue()
        return res


queue = Queue(100)
print(fun(123))
````

1. **What the function is doing**

For every value of num, the function does three important things:

```python
queue.enqueue(num % 10) # take last digit and put into queue

res = fun(num // 10) # recursively process remaining digits

res = res * 10 + queue.dequeue() # while returning, remove from queue
```

The important point I can feel is:

When Python reaches `res = fun(num // 10)`, the current function call **pauses there**. Python first completes the new recursive call.

**Phase 1: Going down into recursion**

fun(123)

**Step 1**: fun(123)
num = 123
queue.enqueue(123 % 10) = 3

Then

```python
res = fun(123 // 10) == fun(12)
```

**At this point fun(123) is paused** this will not go to the next line of the code.

**Step 2**: fun(12)
num = 12
queue.enqueue(12 % 10) = 2

Then

```python

res = fun(12 // 10) == fun(1)
```

and now fun(1) will get called and as of now

```text
fun(123) PAUSED
    |
    V
fun(12)  PAUSED
    |
    V
fun(1)   RUNNING
```

**Step 3**: fun(1)
num = 1
queue.enqueue(1 % 10) = 1

Then

```python

res = fun(1 // 10) == fun(0)
```

and now fun(0) will get called and as of now

```text
fun(123) PAUSED
    |
    V
fun(12)  PAUSED
    |
    V
fun(1)   PAUSED
    |
    V
fun(0)   RUNNING
```

**Step 4**: fun(0)
num = 0
and it will return 0

```text
fun(123) PAUSED
    |
    V
fun(12)  PAUSED
    |
    V
fun(1)   PAUSED
    |
    V
fun(0)  -> return 0
```

**Phase 2: Coming UP from recursion**
execution return first to fun(1) and as fun(1) was waiting for fun(0)

**Step 5: Return to fun(1)**
previously python was waiting
res = fun(0) and fun(0) returned 0

therefore: res = 0

Now Python moves to the next line:

```python
res = res * 10 + queue.dequeue()
```

and the queue currently contains: [3]-> [2]-> [1]

As Queue is FIFO so it will return 3

res = 0 \* 10 + 3 <-- after dequeue call
res = 3

Then, return 3

so, fun(1)-> 3

**Step 6: Return to fun(12)**
res = fun(1)
fun(1) = 3
res = 3 \* 10 + 2 <-- after dequeue call
res = 32

then, return 32

so, fun(12)-> 32

**Step 7: Return to fun(123)**

res= fun(12)
fun(12) = 32
res = 32
res = 32 \* 10 + 1 <-- after dequeue call
res = 321

then, return 321

so, print(fun(123))-> 321

final answer = 321
