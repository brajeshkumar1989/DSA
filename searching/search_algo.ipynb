# **Linear Search**

Brute Forch -> comparing with every item

Start from the starting point till i find the item till the end of the array

We don't need sorted array.

Time complexity is : O(n)

```python
def linear_search(arr, item):
    for i in range(len(arr)):

        if arr[i] == item:
            return i

    return -1
```

```python
arr = [12,34,56,1,67,100,47, 99]

linear_search(arr, 121)
```

# **Binary Search**

Binary Search complexity: nlogn+ logn --> Ammortized concept (Sorting one time only , but searching is done multiple times)

here n < nlogn + logn,

k times , k is a big number
and 




  
 
 


find the middle value = int((lowest_index + highest_index)/2)

ask if that index position value is the value i am looking for, if it is not  then i will check if current value is low or high as compare with the value i am currently seaching , and then the index position i have checked earlier is going to be my higher index value and will repeat the same

```python
# Binary Search

# Sorted Array


def binary_search(arr, low, high, item):
    
    print("low = ", low, "high = ", high, end=" ")
    if low <= high:
        # search          =w
        mid = (low + high) // 2
        print("mid value is: ", arr[mid])
        if arr[mid] == item:
            return mid
        elif arr[mid] > item:
            return binary_search(arr, low, mid - 1, item)
        else:
            return binary_search(arr, mid + 1, high, item)

    else:
        return -1
```

```python
arr = [12,23,34,35,56,77,88,99,100]

print(binary_search(arr, 0, len(arr)-1, 100))
```

### **Sorting**

**When can I say sorting algorithm is stable?**

`A sorting algorithm is said to be stable if two objects with equal keys appear in the same order in sorted output as they appear in the input array to be sorted. Some sorting algorithms are stable by nature like Insertion sort, Merge sort, Bubble sort, etc.And some sorting algorithms are not, like Heap Sort, Quick Sort etc.`



In case of Bubble sort we have a condition arr[i] >arr[i+1] and it means when 2 elements are exact same then both will stay at the same position 

```python
def is_sorted(arr):
    sorted = True

    for i in range(len(arr) - 1):
        if arr[i] > arr[i + 1]:
            sorted = False
    return sorted
```

```python
arr = [1,2,3,9,4,5,6]
is_sorted(arr)
```

In Buble Sort, for a given unsorted array I will keep switching  two adjacent values and complete one pass, and after first pass bigger value will go to last index position of the given unsorted array , then after second pass, second item will be placed in 2nd last position of the array and so on...









In case of 5 items:
    Number of passes = 4
    Number of comparision = 4 + 3 + 2 + 1

In case of N item
    Number of passes = n-1 
    Number of comparision = 1 + 2 + 3...(n-1) = $$\frac{n(n-1)}{2}$$ -> Time complexity O(n) is n^2
    Number of swap = 1 + 2+ 3+...+(n-1)


why this algorithm is known  as bubble sort?

it is like a heavy weighted elements goes down and lighter weighted elements goes up.



```python
def bubble_sort(arr):

    for i in range(len(arr) - 1):
        """internal  loop will run for total number of possble pair if i compare with 1 element to all
        * total comparision will be (n-1) for first pass
        * for 2nd pass i compare with 1 less elements
        """
        for j in range(len(arr) - 1 - i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
        
    return arr

```

Space complexity is 1

```python
arr = [23,12,34,11,100,56,78]

bubble_sort(arr)
```

Adaptive sort: if i pass a sorted array or part of array is sorted already then algorithm should be able to understand if array is already sorted. , and time complexity should be less than algorithm's actual time complexity.



Bubble sort is not adaptive by default.

Can use a new variable flag and change it to 1 if any where swap happens otherwise keep it 0.

So, in best case ,time complexity will be O(n) and worst case it is O(n^2)

And now it is adaptive.




```python
def bubble_sort(arr):

    for i in range(len(arr) - 1):
        flag = 0
        """internal  loop will run for total number of possble pair if i compare with 1 element to all
        * total comparision will be (n-1) for first pass
        * for 2nd pass i compare with 1 less elements
        """
        for j in range(len(arr) - 1 - i):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                flag = 1
        
        if flag == 0:
            break
        
    return arr

```
