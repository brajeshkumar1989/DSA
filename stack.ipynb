# **Stack**

Stack is a data structure

LIFO : Last In First Out

Stack can be build using array or linkedlist.

Stack is open from one end only known as top.

Operation ->

Push - insert element from top.

Pop - reverse operation of Push.

Peak - shows always the top most element

isempty - if stack is empty or not

Size - tells about current size of Stack

All operation from head and tail will have next store as None (when I implement Stack using LinkedList).




```python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None
        
        
```

```python
class Stack:
    def __init__(self):
        self.top = None 
        
    def isempty(self):
        return self.top == None
    
    def push(self, value):
        
        new_node = Node(value)
        
        new_node.next = self.top
        
        self.top = new_node
    
    def peak(self):
        if self.isempty():
            return "Stack Empty"
        
        else:
            return self.top.data
    
    def pop(self):
        if self.isempty():
            return "Stack Empty"
        else:
            data = self.top.data
            self.top = self.top.next
            return data
            
    def size(self):
        if self.isempty():
            return "Stack Empty"
        
        curr = self.top
        count = 0
        while curr != None:
            count += 1
            curr = curr.next
            
        return count
                
    def traverse(self):
        tmp = self.top
        
        while tmp != None:
            print(tmp.data)
            tmp = tmp.next
            

    
    
```

```python
# reverse string


def reverse_string(text):
    s = Stack()

    for i in text:
        s.push(i)

    res = ""

    while not s.isempty():
        res += s.pop()

    print(res)
```

```python
# Text editor
"""
u-> undo
r-> redo

use two stacks.

fill all in u stack and keep r empty

whenever u perform it will add to r stack
and r perform it will add to u stack (original stack)
after complete operation uuuru... print the result
"""

def text_editor(text, pattern):
    u  = Stack()
    r  = Stack()
    
    for i in text:
        u.push(i)
        
    for i in pattern:
        
        if i == 'u':
            data = u.pop()
            r.push(data)
        else:
            data = r.pop()
            u.push(data)
    
    res=""        
    while (not u.isempty()):
        
        res = u.pop() + res
        
    print(res)

```
