# **Implement Queue using Stack**


<!-- @format -->

### Stack properties

- Stack follows LIFO (Last In First Out) principle.
- Push and Pop possible from one end only known as top.

### Queue properties

- Queue follows FIFO (First In First Out) principle.
- It has two end known as front and tail
- front used to dequeue (delete) element.
- rear used to enqueue (insert) element.


<!-- @format -->

- In this problem I am using linkedlist to implement Stack

- Defining a Node class

- Denining Stack class with all required properties


```python
class Node:
    def __init__(self, value):
        self.data = value
        self.next = None
```

```python
class Stack:
    def __init__(self):

        self.top = None

    def isempty(self):

        return self.top == None

    def push(self, value):

        new_node = Node(value)

        new_node.next = self.top

        self.top = new_node

    def peek(self):

        if self.isempty():

            return "Empty Stack."

        return self.top.data

    def pop(self):
        if self.isempty():
            return "Empty Stack."

        data = self.top.data
        self.top = self.top.next

        return data

    def size(self):

        curr = self.top
        count = 0

        while curr != None:
            count += 1
            curr = curr.next
        return count

    def traverse(self):
        if self.isempty():
            return "Empty Stack"
        curr = self.top

        while curr != None:
            print(curr.data, end=" ")
            curr = curr.next
```

- Initializing two empty stacks

- to enqueue use the first stack always
  - enqueue steps
    - push to stack_1

- to dequeue use the second stack always
  - dequeue steps
    - if stack_2 is not empty then simply pop from stack_2

    - if stack_2 is empty pop all items from stack_1 and push to stack_2


```python
class QueueByStack:
    def __init__(self):
        self.stack_1 = Stack()
        self.stack_2 = Stack()

    def enqueue(self, value):
        self.stack_1.push(value)
        print("Current status of stack_1:")
        self.stack_1.traverse()
        print("")
        print("Current status of stack_2:")
        self.stack_2.traverse()

    def dequeue(self):

        # check if stack_1 and stack_2 both are empty
        if self.size() == 0:
            # print("self.size(): ", self.size())
            return "Empty Queue"

        # stack_2 is not empty
        if self.stack_2.top != None:
            data = self.stack_2.pop()
            return f"element {data} has been removed from queue"

        # stack_2 is empty
        while self.stack_1.top != None:

            data = self.stack_1.pop()
            self.stack_2.push(data)

        data = self.stack_2.pop()
        return f"element {data} has been removed from queue"

    def size(self):
        print("stack_1 size: ", self.stack_1.size())
        print("stack_2 size: ", self.stack_2.size())
        return self.stack_1.size() + self.stack_2.size()

    def traverse_queue(self):
        elements = ""
        if self.size() == 0:
            return "Empty Queue"

        if self.stack_2.size() != 0:
            elements1 = ""
            curr = self.stack_2.top
            while curr != None:
                if elements1 == "":
                    elements1 = elements1 + str(curr.data)
                else:
                    elements1 = elements1 + " " + str(curr.data)
                curr = curr.next
            elements = elements1

        if self.stack_1.size() != 0:
            elements2 = ""
            curr = self.stack_1.top
            while curr != None:
                if elements2 == "":
                    elements2 = str(curr.data) + elements2
                else:
                    elements2 = str(curr.data) + " " + elements2
                curr = curr.next
            elements = elements + " " + elements2

        return elements
```

```python
queue_stack = QueueByStack()
```
