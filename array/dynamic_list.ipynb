## **Create List**


```python
import ctypes
```

Pending : negative, indexing, slicing, merge

```python
class MeraList:
    def __init__(self):
        self.size = 1  # total available space
        self.n = 0  # number of total elements

        # create a C type array with size = self.size
        self.A = self.__make_array(self.size)

    # implementation of len method.
    def __len__(self):
        return self.n

    # add new element to list
    def append(self, item):
        # check if vacant, if not resize , yes-> append , A[n]
        if self.n == self.size:
            # resize - double the size
            self.__resize(self.size * 2)

        # append
        self.A[self.n] = item
        self.n = self.n + 1

    # pop method implementation
    def pop(self):
        if self.n == 0:
            return "Empty list"
        print(self.A[self.n - 1])
        self.n = self.n - 1

    # empty the array
    def clear(self):
        self.n = 0
        self.size = 1

    # search the item in a list
    def find(self, item):

        for i in range(self.n):
            if self.A[i] == item:
                return i
        return "ValueError - Not in List"

    # insert new element to specified index position
    def insert(self, pos, item):
        if self.n == self.size:
            self.__resize(self.size * 2)
        for i in range(self.n, pos, -1):
            self.A[i] = self.A[i - 1]
        self.A[pos] = item
        self.n = self.n + 1

    def remove(self, item):
        pos = self.find(item)  # this will give index value

        if type(pos) == int:
            # delete
            self.__delitem__(pos)
        else:
            return pos

    # resize list
    def __resize(self, new_capacity):
        # Create a new array with new capacity
        B = self.__make_array(new_capacity)
        self.size = new_capacity

        # copy the content of A to B
        for i in range(self.n):
            B[i] = self.A[i]

        # reassign A
        self.A = B

    # print implementation
    def __str__(self):
        # [1,2,3]
        result = ""
        for i in range(self.n):
            result = result + str(self.A[i]) + ","
        return "[" + result[:-1] + "]"

    # get element of list index value
    def __getitem__(self, index):
        if 0 <= index < self.n:
            return self.A[index]
        else:
            return "IndexError - Index out of range"

    # Delete item
    def __delitem__(self, pos):
        # delte
        if 0 <= pos < self.n:
            for i in range(pos, self.n - 1):
                self.A[i] = self.A[i + 1]
            self.n = self.n - 1

    def sort(self):
        for i in range(self.n):
            if type(self.A[i]) != int:
                return f"list {self.A} contains element not a integer, value A[{i}]={self.A[i]}"

        for i in range(self.n):
            for j in range(self.n - i - 1):

                if self.A[j] > self.A[j + 1]:
                    self.A[j], self.A[j + 1] = self.A[j + 1], self.A[j]
        return self.A

    def min(self):
        tmp = self.A[0]
        for i in range(1, self.n):
            if tmp > self.A[i]:
                tmp = self.A[i]
        return tmp

    def max(self):
        tmp = self.A[0]
        for i in range(1, self.n):
            if tmp < self.A[i]:
                tmp = self.A[i]
        return tmp

    def sum(self):
        tmp = self.A[0]
        for i in range(1, self.n):
            tmp += self.A[i]

        return tmp

    def extend(self, iterable):
        for item in iterable:
            self.append(item)

    # make array using C type
    def __make_array(self, capacity):
        # creates a C type array(static, referential) with size capacity
        return (capacity * ctypes.py_object)()
```
